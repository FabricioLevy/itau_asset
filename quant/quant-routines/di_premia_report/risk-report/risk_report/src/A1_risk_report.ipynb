```python
import A1_REPORT
import fund_utils as fu
import datetime
import ipywidgets as widgets

from IPython.display import HTML

HTML('''<script>
code_show=true; 
function code_toggle() {
 if (code_show){
 $('div.input').hide();
 } else {
 $('div.input').show();
 }
 code_show = !code_show
} 
$( document ).ready(code_toggle);
</script>
<form action="javascript:code_toggle()"><input type="submit" value="Click here to toggle on/off the raw code."></form>''')
```

```python
dt_p = widgets.DatePicker(
    description='Date to Run Risk Report: ',
    disabled=False,
    value = datetime.date.today(),
    style = {'description_width': 'initial'},
    layout=widgets.Layout(width='300px', height='30px')
)

fund_dp = widgets.Dropdown(
    options=['A1 Master FIM', 'A1 Orange FIM', 'A1 FIFE BEM'],
    value='A1 Master FIM',
    description='Fund: ',
    disabled=False,
)

btn_go = widgets.Button(
    description='Run Risk Report',
    disabled=False,
    button_style='', # 'success', 'info', 'warning', 'danger' or ''
    tooltip='Run',
    icon='check'
)

output = widgets.Output(layout={'border': '10px solid grey'})

display(dt_p,fund_dp, btn_go, output)

def on_button_clicked(btn_go):
    dt, str_fund = dt_p.value, fund_dp.value
    with output:
        output.clear_output()
        try:
            nav = fu.get_NAV(dt, str_fund)
        except Exception as err:
            print("ERROR: {}".format(err))
            return False
        
        print("Running report...")
        if A1_REPORT.main(dt, str_fund, nav):
            print("Report saved!")

btn_go.on_click(on_button_clicked)
```
